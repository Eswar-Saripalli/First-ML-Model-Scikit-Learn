## Step 1: Import Required Libraries

Import all the required Python libraries for data analysis, visualization, and machine learning.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
```

## Dataset Information

The Iris dataset is a classic machine learning dataset containing 150 flower samples with four features:
- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

The target variable contains three classes:
- Setosa
- Versicolor
- Virginica

## Step 2: Load the Iris Dataset

Load the Iris dataset from the scikit-learn library.

```python
iris = load_iris()
```

## Step 3: Create the DataFrame

The Iris dataset contains four input features:
- Sepal Length (cm)
- Sepal Width (cm)
- Petal Length (cm)
- Petal Width (cm)

The target variable contains three flower species:
- Setosa
- Versicolor
- Virginica

Convert the dataset into a Pandas DataFrame for easier analysis.

```python
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
df.head()
```

## Step 4: Check the Dataset Shape

Display the number of rows and columns in the dataset.

```python
df.shape
```

## Step 5: Display Dataset Information

Display the dataset information, including column names, data types, and missing values.

```python
df.info()
```

## Step 6: Display Statistical Summary

Generate descriptive statistics such as count, mean, standard deviation, minimum, maximum, and quartile values for the numerical features.

```python
df.describe()
```

## Step 7: Add the Target Variable

Add the target variable (species) to the DataFrame for machine learning.

```python
df["species"] = iris.target
```

## Step 8: Preview the Updated Dataset

Display the first five rows after adding the target column.

```python
df.head()
```

## Step 9: Check Class Distribution

Display the number of samples available in each species class.

```python
df["species"].value_counts()
```

## Step 10: Prepare Features and Target

Separate the input features (X) and the target variable (y), then import the train-test split function.

```python
from sklearn.model_selection import train_test_split

x = df.drop("species", axis=1)
y = df["species"]
```

## Step 11: Split the Dataset

Split the dataset into training and testing sets using an 80:20 ratio.

```python
x_train, x_test, y_train, y_test = train_test_split(
   x, y, test_size=0.2, random_state=42
)
```

## Step 12: Train the Logistic Regression Model

Import the Logistic Regression algorithm, create the model, and train it using the training dataset.

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(random_state=42)
model.fit(x_train, y_train)
```

## Step 13: Make Predictions

Use the trained Logistic Regression model to predict the classes of the test dataset.

```python
y_pred = model.predict(x_test)
```

## Step 14: Evaluate the Logistic Regression Model

Calculate the accuracy of the Logistic Regression model using the test dataset.

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

## Step 15: Generate the Confusion Matrix

Generate the confusion matrix to evaluate the performance of the Logistic Regression model.

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

## Step 16: Generate the Classification Report

Generate the classification report to evaluate the precision, recall, F1-score, and overall performance of the Logistic Regression model.

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

## Step 17: Train the Random Forest Model

Import the Random Forest Classifier, train the model using the training dataset, and evaluate its accuracy.

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(x_train, y_train)
rf_pred = rf_model.predict(x_test)
rf_accuracy = accuracy_score(y_test, rf_pred)
print("Random Forest Accuracy:", rf_accuracy)
```

## Step 18: Compare Model Performance

Compare the accuracy of the Logistic Regression and Random Forest models.

```python
comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy": [accuracy, rf_accuracy]
})

comparison
```

## Step 19: Visualize Model Accuracy

Create a bar chart to compare the accuracy of both machine learning models.

```python
plt.figure(figsize=(6,4))
plt.bar(comparison["Model"], comparison["Accuracy"])
plt.title("Model Accuracy Comparison")
plt.xlabel("Models")
plt.ylabel("Accuracy")
plt.ylim(0.9, 1.05)
plt.show()
```

## Step 20: Plot the Confusion Matrix

Visualize the confusion matrices of both models to evaluate their prediction performance.

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_estimator(model, x_test, y_test)
plt.title("Logistic Regression")
plt.show()

ConfusionMatrixDisplay.from_estimator(rf_model, x_test, y_test)
plt.title("Random Forest")
plt.show()
```
